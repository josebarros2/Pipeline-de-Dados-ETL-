```python
import requests
import pandas as pd
import sqlite3

print("Iniciando extração de dados...")
url = "https://brasilapi.com.br/api/feriados/v1/2025"
response = requests.get(url)

if response.status_code == 200:
    dados = response.json()
    print("Dados extraídos com sucesso!")

    df = pd.DataFrame(dados)

    df['date'] = pd.to_datetime(df['date'])

    df['data_processamento'] = pd.Timestamp.now()

    print("Dados transformados:")
    print(df.head())

    conn = sqlite3.connect('feriados_brasil.db')
    df.to_sql('tb_feriados', conn, if_exists='replace', index=False)

    print("\nSucesso! Os dados foram salvos no banco de dados SQL 'feriados_brasil.db'.")
    conn.close()
else:
    print(f"Erro ao acessar API: {response.status_code}")
```

```python

conn = sqlite3.connect('feriados_brasil.db')

query = """
SELECT name, date
FROM tb_feriados
WHERE date <= '2025-06-30'
ORDER BY date ASC
"""

df_semestre = pd.read_sql_query(query, conn)

print("Feriados do 1º Semestre (Consultados via SQL):")
print(df_semestre)

conn.close()
```
